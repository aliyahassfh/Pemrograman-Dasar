```python
# Soal 9.

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]

jumlah_tidak_valid = 0

for i, nilai in enumerate(data, start=1):
    try:
        vibrasi = float(nilai)

    except ValueError:
        print(f"Unit {i}: data tidak valid")
        jumlah_tidak_valid += 1
        continue

    if vibrasi > 2.0:
        print(f"Warning: Unit {i} vibrasi tinggi!")

    if vibrasi > 3.0:
        print("KRITIS: pemeriksaan dihentikan")
        break

print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

**Cara kerja program dan fungsi kode**

Soal 9
*   data = [...] digunakan untuk menyimpan data pembacaan vibrasi dari lima unit mesin dalam bentuk teks.
*   jumlah_tidak_valid = 0 digunakan sebagai penghitung data yang tidak valid.
*   for i, nilai in enumerate(data, start=1) digunakan untuk memeriksa setiap data satu per satu. Nomor unit dimulai dari 1.
*   try digunakan untuk mencoba mengubah data teks menjadi tipe float.
*   float(nilai) digunakan untuk mengubah data vibrasi menjadi angka desimal.
*   except ValueError digunakan untuk menangani data yang tidak dapat diubah menjadi angka, seperti "ERR".
*   jumlah_tidak_valid += 1 digunakan untuk menambah penghitung jika ditemukan data tidak valid.
*   continue digunakan untuk melewati data yang tidak valid dan melanjutkan pemeriksaan ke unit berikutnya.
*   if vibrasi > 2.0 digunakan untuk memeriksa apakah nilai vibrasi melebihi 2.0 dan menampilkan peringatan.
*  if vibrasi > 3.0 digunakan untuk memeriksa apakah nilai vibrasi melebihi 3.0.
*   break digunakan untuk menghentikan perulangan jika ditemukan vibrasi yang melebihi 3.0.
*   print(f"Jumlah data tidak valid: {jumlah_tidak_valid}") digunakan untuk menampilkan jumlah data tidak valid setelah perulangan selesai.














```python

# Nama    : Aliyah Nur Assyfah
# NIM     : 4232501034
# Tanggal : 10 November 2026
# Mata Kuliah : Pemrograman Dasar

# Challenge: Pemantauan Konsumsi Listrik Harian
# Program ini digunakan untuk memantau konsumsi listrik selama 7 hari.
# Program menghitung total, rata-rata, dan status konsumsi listrik.

konsumsi_listrik = []

# Memasukkan data konsumsi listrik selama 7 hari
for hari in range(1, 8):
    while True:
        try:
            konsumsi = float(input(f"Masukkan konsumsi listrik hari ke-{hari} (kWh): "))

            if konsumsi < 0:
                print("Konsumsi tidak boleh negatif!")
            else:
                konsumsi_listrik.append(konsumsi)
                break

        except ValueError:
            print("Input tidak valid! Masukkan angka.")

# Menghitung total dan rata-rata konsumsi listrik
total = sum(konsumsi_listrik)
rata_rata = total / len(konsumsi_listrik)

print("\n=== HASIL PEMANTAUAN LISTRIK ===")

# Menampilkan konsumsi dan status setiap hari
for i, konsumsi in enumerate(konsumsi_listrik, start=1):
    print(f"Hari ke-{i}: {konsumsi} kWh")

    if konsumsi < 20:
        print("Status: Hemat")
    elif konsumsi <= 30:
        print("Status: Normal")
    else:
        print("Status: Tinggi")

# Menampilkan hasil perhitungan
print(f"\nTotal konsumsi listrik: {total:.2f} kWh")
print(f"Rata-rata konsumsi listrik: {rata_rata:.2f} kWh")

# Menentukan status rata-rata konsumsi selama 7 hari
if rata_rata < 20:
    print("Kesimpulan: Konsumsi listrik tergolong hemat.")
elif rata_rata <= 30:
    print("Kesimpulan: Konsumsi listrik tergolong normal.")
else:
    print("Kesimpulan: Konsumsi listrik tergolong tinggi.")
```

**Cara kerja program dan fungsi kode**

Soal challenge
*   konsumsi_listrik = [] digunakan untuk membuat list kosong sebagai tempat menyimpan data konsumsi listrik selama 7 hari.
*   for hari in range(1, 8) digunakan untuk melakukan perulangan sebanyak 7 kali sesuai jumlah hari pengamatan.
*   while True digunakan agar pengguna dapat memasukkan ulang data jika terjadi kesalahan input.
*   try-except ValueError digunakan untuk menangani kesalahan ketika pengguna memasukkan data yang bukan angka.
*   if konsumsi < 0 digunakan untuk memastikan nilai konsumsi listrik tidak negatif.
*   append(konsumsi) digunakan untuk menambahkan data konsumsi listrik ke dalam list.
*   sum(konsumsi_listrik) digunakan untuk menghitung total konsumsi listrik selama 7 hari.
*   len(konsumsi_listrik) digunakan untuk menghitung jumlah data, sedangkan pembagian total dengan jumlah data digunakan untuk mendapatkan rata-rata konsumsi listrik.
*   enumerate(konsumsi_listrik, start=1) digunakan untuk menampilkan data konsumsi beserta nomor harinya.
*   if-elif-else digunakan untuk menentukan status konsumsi listrik:
- Kurang dari 20 kWh = Hemat.
- 20 sampai 30 kWh = Normal.
- Lebih dari 30 kWh = Tinggi.
*   print() digunakan untuk menampilkan hasil pemantauan, total konsumsi, rata-rata, dan kesimpulan.

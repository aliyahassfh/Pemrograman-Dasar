```python
# Program Jajan di Kantin

jumlah_gorengan = 3
harga_gorengan = 2500
jumlah_es_teh = 2
harga_es_teh = 4000
uang_dibayar = 20000

total_gorengan = jumlah_gorengan * harga_gorengan
total_es_teh = jumlah_es_teh * harga_es_teh
total_belanja = total_gorengan + total_es_teh
kembalian = uang_dibayar - total_belanja
uang_cukup = uang_dibayar >= total_belanja

#Output
print("=== Jajan di Kantin ===")
print("Harga gorengan   : Rp", total_gorengan)
print("Harga es teh     : Rp", total_es_teh)
print("Total belanja    : Rp", total_belanja)
print("Uang dibayar     : Rp", uang_dibayar)
print("Uang kembalian   : Rp", kembalian)
print("Apakah uang cukup?", uang_cukup)
```

```python
# Durasi Film Marathon

total_menit = 275

jam_total = total_menit // 60
menit_sisa = total_menit % 60
jam_desimal = total_menit / 60
cek_rentang = jam_desimal > 3 and jam_desimal < 5

#Output
print(f"Durasi film: {jam_total} jam {menit_sisa} menit")
print(f"Total durasi dalam jam desimal: {jam_desimal:.2f}")
print(f"Apakah durasi lebih dari 3 jam dan kurang dari 5 jam? {cek_rentang}")
```

```python

# Nama  : Aliyah Nur Assyfah
# NIM   : 4232501034
# Tujuan: Menghitung total belanja, diskon, kembalian,
#         iuran per orang, sisa pembagian, dan mengecek
#         apakah uang yang dimiliki cukup.

jumlah_buku = 7
harga_buku = 18500
diskon = 10 / 100
uang_budi = 150000
jumlah_orang = 4

# Nomor 1: Menghitung total belanja dan diskon
total_belanja = jumlah_buku * harga_buku
besar_diskon = total_belanja * diskon
total_bayar = total_belanja - besar_diskon

print("1. Total Belanja")
print("Total belanja: Rp", total_belanja)
print("Besar diskon: Rp", int(besar_diskon))
print("Jumlah yang harus dibayar: Rp", int(total_bayar))

# Nomor 2: Menghitung uang kembalian
kembalian = uang_budi - total_bayar

print("\n2. Uang Kembalian")
print("Uang kembalian: Rp", int(kembalian))

# Nomor 3: Menghitung iuran dan sisa pembagian
iuran = int(total_bayar) // jumlah_orang
sisa = int(total_bayar) % jumlah_orang

print("\n3. Iuran Empat Orang")
print("Iuran per orang: Rp", iuran)
print("Sisa pembagian: Rp", sisa)

# Nomor 4: Memeriksa apakah uang Budi cukup
uang_cukup = uang_budi >= total_bayar

print("\n4. Pemeriksaan Uang")
print("Apakah uang Budi cukup?", uang_cukup)
```

```python

# Nama  : Aliyah Nur Assyfah
# NIM   : 4232501034
# Tujuan: Menghitung energi listrik yang dihasilkan PLTS,
#         nilai listrik per bulan, kekurangan energi,
#         dan mengecek kecukupan energi rumah.

jumlah_panel = 12
daya_panel = 450
penyinaran = 5
efisiensi = 0.8
kebutuhan = 25
tarif_listrik = 1444.70

# Nomor 1: Menghitung energi harian
energi_harian = jumlah_panel * daya_panel * penyinaran * efisiensi / 1000

# Nomor 2: Menghitung energi bulanan dan nilai listrik
energi_bulanan = energi_harian * 30
nilai_listrik = energi_bulanan * tarif_listrik

# Nomor 3: Menghitung kekurangan energi harian
kekurangan = kebutuhan - energi_harian

# Nomor 4: Memeriksa kecukupan energi
mencukupi = energi_harian >= kebutuhan

print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(kekurangan, 1))
print("Mencukupi?          :", mencukupi)
```

```python

# Nama  : Aliyah Nur Assyfah
# NIM   : 4232501034
# Tujuan: Menghitung energi generator, kebutuhan dan biaya
#         solar, kecukupan tangki, serta rugi daya kabel.

daya_generator = 200
waktu_operasi = 8
konsumsi_solar = 0.25
harga_solar = 13500
kapasitas_tangki = 1000
arus = 20
hambatan = 0.5

# Nomor 1: Menghitung energi harian
energi = daya_generator * waktu_operasi

# Nomor 2: Menghitung kebutuhan dan biaya solar
solar_harian = energi * konsumsi_solar
biaya_solar = solar_harian * harga_solar

# Nomor 3: Menghitung kecukupan tangki dan sisa solar
hari_penuh = kapasitas_tangki // solar_harian
sisa_solar = kapasitas_tangki % solar_harian

# Nomor 4: Menghitung rugi daya kabel
rugi_kabel = arus ** 2 * hambatan

# Nomor 5: Memeriksa apakah rugi kabel tergolong tinggi
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("Energi (kWh)     :", energi)
print("Solar (liter)    :", float(solar_harian))
print("Biaya (Rp)       :", float(biaya_solar))
print("Tangki cukup     :", float(hari_penuh), "hari")
print("Sisa solar (L)   :", float(sisa_solar))
print("Rugi kabel (W)   :", float(rugi_kabel))
print("Rugi tinggi?     :", rugi_tinggi)
```

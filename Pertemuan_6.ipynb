```python
# Nama  : Aliyah Nur Assyfah
# NIM   : 4232501034
# Tujuan: Menghitung faktorial bilangan bulat dari 1 sampai 5
#         menggunakan percabangan if, elif, dan else.
# Soal 1

n = int(input("Masukkan bilangan bulat (1-5): "))

if n < 1 or n > 5:
    print("Input tidak valid. Masukkan bilangan antara 1 sampai 5.")
elif n == 1:
    print("1! = 1")
elif n == 2:
    print("2! =", 2 * 1)
elif n == 3:
    print("3! =", 3 * 2 * 1)
elif n == 4:
    print("4! =", 4 * 3 * 2 * 1)
else:
    print("5! =", 5 * 4 * 3 * 2 * 1)
```

```python
# Nama  : Aliyah Nur Assyfah
# NIM   : 4232501034
# Tujuan: Menghitung skor kualitas daya dan mengklasifikasikan
#         kualitas daya berdasarkan faktor daya dan efisiensi distribusi.
# Soal 2

faktor_daya = float(input("Masukkan faktor daya (0.0 - 1.0): "))
efisiensi = float(input("Masukkan efisiensi distribusi (%): "))

skor = faktor_daya * efisiensi

print(f"Skor kualitas daya: {skor:.2f}")

if skor >= 85 and faktor_daya >= 0.90:
    klasifikasi = "PRIMA — sistem dalam kondisi optimal"
elif skor >= 70 and faktor_daya >= 0.80:
    klasifikasi = "BAIK — sistem memadai, pantau rutin"
elif skor >= 50 or faktor_daya >= 0.75:
    klasifikasi = "CUKUP — ada ruang perbaikan"
else:
    klasifikasi = "BURUK — perlu perbaikan segera"

print(f"Faktor daya {faktor_daya}, efisiensi {efisiensi}% → Klasifikasi: {klasifikasi}")
```

```python
# Nama  : Aliyah Nur Assyfah
# NIM   : 4232501034
# Tujuan: Menentukan mode operasi PLTS berdasarkan iradiasi
#         matahari dan kapasitas baterai yang tersisa.
# Soal 3

iradiasi = float(input("Masukkan iradiasi matahari (W/m2): "))
soc = float(input("Masukkan SoC baterai (%): "))

daya_panel = iradiasi * 0.18 * 10

print(f"Estimasi daya panel: {daya_panel:.1f} W")

if iradiasi > 600 and soc < 80:
    mode = "CHARGING — prioritaskan pengisian baterai"
elif iradiasi > 600 and soc >= 80:
    mode = "GRID-EXPORT — ekspor kelebihan daya ke jaringan"
elif iradiasi > 200 and soc >= 20:
    mode = "NORMAL — operasi mandiri"
elif iradiasi <= 200 and soc >= 50:
    mode = "BATTERY-ONLY — andalkan baterai"
else:
    mode = "EMERGENCY — kurangi beban segera"

print(f"Iradiasi {iradiasi} W/m2, SoC {soc}% → Mode: {mode}")
```

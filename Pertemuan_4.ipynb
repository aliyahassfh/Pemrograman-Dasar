```python
# konversi waktu
total_detik = int(input("Masukkan total detik: "))

jam = total_detik // 3600
sisa_detik = total_detik % 3600
menit = sisa_detik // 60
detik = sisa_detik % 60

print(f"Hasil Konversi: {jam} Jam, {menit} Menit, {detik} Detik")
```

```python
nilai_mahasiswa = float(input("Masukkan nilai mahasiswa: "))
standard_kelulusan = float(input("Masukkan standar kelulusan: "))

apakah_lulus = nilai_mahasiswa >= standard_kelulusan
perlu_remedial = nilai_mahasiswa < standard_kelulusan

print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standard_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
tugas = float(input("Masukkan nilai tugas: "))
uts = float(input("Masukkan nilai UTS: "))
uas = float(input("Masukkan nilai UAS: "))
praktikum = float(input("Masukkan nilai praktikum: "))
kuis = float(input("Masukkan nilai kuis: "))

nilai_mahasiswa = (
    (0.2 * tugas) +
    (0.2 * uts) +
    (0.3 * uas) +
    (0.2 * praktikum) +
    (0.1 * kuis)
)

standard_kelulusan = float(input("Masukkan standar kelulusan: "))

apakah_lulus = nilai_mahasiswa >= standard_kelulusan
perlu_remedial = nilai_mahasiswa < standard_kelulusan

print(f"Nilai Mahasiswa: {nilai_mahasiswa}")
print(f"Standar Kelulusan: {standard_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
belanja = float(input("Masukkan total belanja: "))

diskon = (0.1 * belanja) if belanja > 500000 else 0

total_harga = belanja - diskon

print("Total harga setelah diskon =", total_harga)
```

```python
# Latihan 1

tahun = int(input("Masukkan tahun:"))

tahun_kabisat = (tahun % 4 == 0 and tahun % 100 != 0) or (tahun % 400 == 0)

if tahun_kabisat:
  print(tahun, "adalah tahun kabisat")
else:
  print(tahun, "bukan tahun kabisat")
```

```python
# Latihan 2
# Program menghitung luas persegi panjang

def hitung_luas():
  panjang = float(input("Masukkan panjang:"))
  lebar = float(input("Masukkan lebar:"))

  luas = panjang * lebar

  print("Panjang =", panjang)
  print("Lebar =", lebar)
  print("Luas persegi panjang =", luas)

hitung_luas()
```
